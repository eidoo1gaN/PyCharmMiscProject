```python
import matplotlib.pyplot as plt
import numpy as np
```

# Basic Trajectory

```python
#angle for this example is in degrees
theta = np.radians(30)
#initial velocity
u = 100
#gravity or generally acceleration
g = 9.8
#root equation
root = (2*(u**2)*(np.cos(theta)**2)*(np.tan(theta)))/g

x = np.linspace(0, root)
#trajectory formula
y = (-(g*(x**2))/((2*(u**2))*(np.cos(theta)**2)))+(x*np.tan(theta))
```

```python
fig, ax = plt.subplots()
ax.autoscale(enable=False, axis='y')
ax.set_ylim(bottom=0, top=200)

ax.plot(x, y)

plt.style.use("default")
ax.grid()

plt.xlabel("Distance (m)")
plt.ylabel("Height (m)")



plt.show()
```
