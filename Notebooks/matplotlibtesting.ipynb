```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pds

plt.style.use("default")
```

```python
x = np.linspace(0, 20)
plt.figure(figsize=(20, 10))


y1 = np.sin(x)
y2 = (1 + x * np.random.randn())/(np.cos(x))

plt.scatter(x, y1, alpha=0.7)
plt.plot(x, y2)
plt.xlabel("Time")
plt.ylabel("Boringness")
plt.show()
```
